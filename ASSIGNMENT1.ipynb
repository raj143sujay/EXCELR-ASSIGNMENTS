EXERCISE 1 
PRIME NUMBER

```python
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

Exercise 2: Product of Random Numbers

```python
import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

print("Number 1:", num1)
print("Number 2:", num2)

answer = eval(input("Enter the product of these numbers: "))

if answer == num1 * num2:
    print("Correct Answer!")
else:
    print("Wrong Answer!")
    print("Correct product is:", num1 * num2)
```

EXERCISE 3: Squares of Even/Odd Numbers

```python
print("Squares of even numbers from 100 to 200:")

for i in range(100, 201):
    if i % 2 == 0:
        print(i, "=", i * i)
```

Exercise 4:
✅ Problem Statement (Question): Hospital Billing System


```python
# Charges
room_charge_per_day = 2000
doctor_fee = 1500
lab_test_cost = 300

# Patient details
days_stayed = 4
number_of_tests = 3
medicine_charges = 2400

# Calculations
room_charges = room_charge_per_day * days_stayed
lab_charges = number_of_tests * lab_test_cost

total_bill = room_charges + doctor_fee + lab_charges + medicine_charges

# Discount
if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

final_bill = total_bill - discount

# Output
print("Room Charges:", room_charges)
print("Doctor Fee:", doctor_fee)
print("Lab Charges:", lab_charges)
print("Medicine Charges:", medicine_charges)
print("Total Bill:", total_bill)
print("Discount:", discount)
print("Final Bill:", final_bill)
```
