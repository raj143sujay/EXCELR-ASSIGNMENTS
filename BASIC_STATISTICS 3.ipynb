```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("sales_data_with_discounts.csv")

# Display first 5 rows
print("First 5 Rows:")
print(df.head())

# -----------------------------
# DESCRIPTIVE ANALYTICS
# -----------------------------

# Select numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns

print("\nNumerical Columns:")
print(numerical_cols)

# Mean
print("\nMean:")
print(df[numerical_cols].mean())

# Median
print("\nMedian:")
print(df[numerical_cols].median())

# Mode
print("\nMode:")
print(df[numerical_cols].mode().iloc[0])

# Standard Deviation
print("\nStandard Deviation:")
print(df[numerical_cols].std())

# Full Statistical Summary
print("\nStatistical Summary:")
print(df[numerical_cols].describe())

# -----------------------------
# HISTOGRAMS
# -----------------------------

for col in numerical_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=10)
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# -----------------------------
# BOXPLOTS
# -----------------------------

for col in numerical_cols:
    plt.figure(figsize=(6,4))
    plt.boxplot(df[col])
    plt.title(f'Boxplot of {col}')
    plt.ylabel(col)
    plt.show()

# -----------------------------
# CATEGORICAL COLUMNS
# -----------------------------

categorical_cols = df.select_dtypes(include='object').columns

print("\nCategorical Columns:")
print(categorical_cols)

# Bar Charts
for col in categorical_cols:
    plt.figure(figsize=(8,4))
    df[col].value_counts().plot(kind='bar')
    plt.title(f'Bar Chart of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.show()

# -----------------------------
# CONCLUSION
# -----------------------------

print("\nConclusion:")
print("1. Descriptive statistics calculated successfully.")
print("2. Histograms show the distribution of numerical variables.")
print("3. Boxplots help identify outliers.")
print("4. Bar charts show the frequency distribution of categorical variables.")
print("5. Dataset is ready for further analysis and modeling.")
```
