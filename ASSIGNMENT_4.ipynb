```python
import pandas as pd

# Load datasets
patients = pd.read_csv("Patient_Data.csv")
billing = pd.read_csv("Billing_Data.csv")

# 1. Summary using info()
print("PATIENT DATA INFO")
patients.info()

# 2. Select billing-related columns
billing_cols = patients[['PatientID', 'Department', 'Doctor', 'BillAmount']]
print("\nBilling Columns:")
print(billing_cols)

# 3. Drop administrative columns
patients_clean = patients.drop(
    columns=['ReceptionistID', 'CheckInTime']
)
print("\nAfter Dropping Columns:")
print(patients_clean)

# 4. Fill missing bill amounts with mean
mean_bill = patients_clean['BillAmount'].mean()
patients_clean['BillAmount'] = patients_clean['BillAmount'].fillna(mean_bill)

# Groupby Department
dept_bill = patients_clean.groupby('Department')['BillAmount'].sum()

print("\nTotal Bill Amount per Department:")
print(dept_bill)

# 5. Remove duplicate PatientID records
patients_clean = patients_clean.drop_duplicates(subset='PatientID')

print("\nAfter Removing Duplicates:")
print(patients_clean)

# 6. Merge billing dataset
merged_df = pd.merge(
    patients_clean,
    billing,
    on='PatientID',
    how='left'
)

print("\nMerged Dataset:")
print(merged_df)

# 7. New patients dataframe
new_patients = pd.DataFrame({
    'PatientID':[106,107],
    'Name':['John','Mary'],
    'Department':['Cardiology','Neurology'],
    'Doctor':['Dr. Kumar','Dr. Rao'],
    'BillAmount':[7000,6500]
})

# Row-wise concatenate
final_df = pd.concat(
    [patients_clean, new_patients],
    ignore_index=True
)

print("\nAfter Adding New Patients:")
print(final_df)

# 8. Column-wise concatenate
extra_cols = pd.DataFrame({
    'InsuranceCovered':[1000,1500,1200,800,900,1100,1000],
    'FinalAmount':[5000,3500,6200,4200,4000,6000,5500]
})

final_df = pd.concat(
    [final_df, extra_cols],
    axis=1
)

print("\nFinal Dataset:")
print(final_df)
```
