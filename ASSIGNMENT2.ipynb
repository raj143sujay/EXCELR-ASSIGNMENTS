Section A: Data Structures & Control Structures
1. List Operations

```python
nums = [10, 20, 30, 40, 50]

nums.append(60)
print("append:", nums)

nums.extend([70, 80])
print("extend:", nums)

nums.insert(2, 25)
print("insert:", nums)

nums.remove(25)
print("remove:", nums)

nums.pop()
print("pop:", nums)

print("index of 40:", nums.index(40))

print("count of 20:", nums.count(20))

nums.sort()
print("sort:", nums)

nums.reverse()
print("reverse:", nums)

nums.clear()
print("clear:", nums)
```

2. Tuple

```python
students = ("Ravi", "Priya", "Kiran")

students_list = list(students)
students_list[1] = "Anjali"

students = tuple(students_list)

print(students)
```

3. Set with Duplicates

```python
numbers = {1, 2, 3, 4, 4, 5, 2, 1}

print(numbers)
```

4. Dictionary

```python
person = {
    "name": "Raj",
    "age": 20,
    "city": "Nellore"
}

person["age"] = 21
person["email"] = "raj@gmail.com"

print(person)
```

5. Voting Eligibility

```python
age = 19

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

6. Grade Calculator

```python
marks = 82

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 50:
    print("C")
else:
    print("Fail")
```

7. Positive, Even, Zero or Negative

```python
num = 12

if num > 0:
    print("Positive Number")

    if num % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")

elif num == 0:
    print("Zero")

else:
    print("Negative Number")
```

Section C: Exploring Pandas
11. DataFrame Operations

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Raj", "Anu", "Kiran", "Priya", "Ravi",
             "Sita", "Arun", "Divya", "Raj", "Kiran"],

    "Age": [20, 22, 21, 23, 24,
            25, 22, 21, 20, 21],

    "City": ["Nellore", "Chennai", "Hyderabad", "Bangalore",
             "Mumbai", "Delhi", "Pune", "Kolkata",
             "Nellore", "Hyderabad"],

    "Salary": [25000, 30000, np.nan, 35000,
               40000, 28000, np.nan, 32000,
               25000, np.nan]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nInfo:")
print(df.info())

print("\nDescribe:")
print(df.describe())

print("\nName and City Columns:")
print(df[["Name", "City"]])

df = df.drop("City", axis=1)

mean_salary = df["Salary"].mean()
df["Salary"] = df["Salary"].fillna(mean_salary)

df = df.drop_duplicates()

print("\nFinal DataFrame:")
print(df)
```
