```python
import numpy as np
import scipy.stats as stats

# --- Part 1: Hypothesis Testing ---
print("="*50)
print("PART 1: HYPOTHESIS TESTING REPORT")
print("="*50)

# Given data
n = 25                       # Sample size
sample_mean = 3050           # Sample mean weekly cost
X_mean = 600                 # Mean of units produced
X_std = 25                   # Standard deviation of units produced
alpha = 0.05                 # Significance level

# 1. State the Hypotheses
print("Step 1: Hypotheses Statement")
print("Null Hypothesis (H0): mu <= $4,000 (Weekly operating costs have not increased)")
print("Alternative Hypothesis (H1): mu > $4,000 (Weekly operating costs are higher)")
print("-" * 30)

# 2. Calculate the Theoretical Mean (mu) and Population Standard Deviation (sigma)
# Model: W = 1000 + 5X
mu = 1000 + 5 * X_mean
sigma = 5 * X_std

print(f"Step 2: Calculated Parameters")
print(f"Theoretical Mean Weekly Cost (mu): ${mu}")
print(f"Population Standard Deviation (sigma): {sigma}")

# Calculate the Z-test statistic
# Formula: z = (sample_mean - mu) / (sigma / sqrt(n))
z_stat = (sample_mean - mu) / (sigma / np.sqrt(n))
print(f"Calculated Z-Test Statistic: {z_stat:.4f}")
print("-" * 30)

# 3. Determine the Probability (p-value) and Critical Value
p_value = 1 - stats.norm.cdf(z_stat)
z_critical = stats.norm.ppf(1 - alpha)

print("Step 3 & 4: Probability and Decision Metrics")
print(f"P-value: {p_value:.4f}")
print(f"Critical Z-value (at alpha={alpha}): {z_critical:.4f}")
print("-" * 30)

# 4. Make a Decision
print("Step 5: Decision & Conclusion")
if z_stat > z_critical:
    print(f"Decision: Reject the Null Hypothesis (H0) because Z-statistic ({z_stat:.4f}) > Critical Z-value ({z_critical:.4f}).")
    print("Conclusion: There is strong evidence to support the restaurant owners' claim that the weekly operating costs are higher than the model suggests.")
else:
    print(f"Decision: Fail to Reject the Null Hypothesis (H0) because Z-statistic ({z_stat:.4f}) <= Critical Z-value ({z_critical:.4f}).")
    print("Conclusion: There is not enough evidence to support the restaurant owners' claim.")
print("="*50 + "\n")
```

```python
# --- Part 2: Estimation and Confidence Intervals ---
print("="*50)
print("PART 2: CONFIDENCE INTERVALS REPORT")
print("="*50)

# Data
durability_data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
n_ci = len(durability_data)
confidence_level = 0.99

# Sample descriptive statistics
x_bar = np.mean(durability_data)
s = np.std(durability_data, ddof=1) # Sample standard deviation (Bessel's correction)

print(f"Sample Metrics:")
print(f"Sample Size (n): {n_ci}")
print(f"Sample Mean (x_bar): {x_bar:.4f} million characters")
print(f"Sample Standard Deviation (s): {s:.4f} million characters")
print("-" * 30)

# Task a: 99% Confidence Interval Using Sample Standard Deviation (t-distribution)
print("Task a: 99% CI using Sample Standard Deviation (t-distribution)")
print("Rationale: The t-distribution is used because the population standard deviation is unknown ")
print("and the sample size is small (n = 15 < 30).")

df = n_ci - 1 # Degrees of freedom
t_critical = stats.t.ppf((1 + confidence_level) / 2, df)
margin_of_error_t = t_critical * (s / np.sqrt(n_ci))
ci_t_lower = x_bar - margin_of_error_t
ci_t_upper = x_bar + margin_of_error_t

print(f"Critical t-value: {t_critical:.4f}")
print(f"99% Confidence Interval: ({ci_t_lower:.4f}, {ci_t_upper:.4f}) million characters")
print("-" * 30)

# Task b: 99% Confidence Interval Using Known Population Standard Deviation (z-distribution)
print("Task b: 99% CI using Known Population Standard Deviation (z-distribution)")
sigma_known = 0.2

z_critical_ci = stats.norm.ppf((1 + confidence_level) / 2)
margin_of_error_z = z_critical_ci * (sigma_known / np.sqrt(n_ci))
ci_z_lower = x_bar - margin_of_error_z
ci_z_upper = x_bar + margin_of_error_z

print(f"Known Population Standard Deviation (sigma): {sigma_known}")
print(f"Critical z-value: {z_critical_ci:.4f}")
print(f"99% Confidence Interval: ({ci_z_lower:.4f}, {ci_z_upper:.4f}) million characters")
print("="*50)
```
