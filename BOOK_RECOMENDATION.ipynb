## 1. Load Data and Initial Inspection

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
users_df = pd.read_csv('/content/Users.csv')
books_df = pd.read_csv('/content/Books.csv')
ratings_df = pd.read_csv('/content/Ratings.csv')
```

### Users DataFrame

```python
print('Users DataFrame - Head:')
display(users_df.head())
print('\nUsers DataFrame - Info:')
display(users_df.info())
print('\nUsers DataFrame - Describe:')
display(users_df.describe())
```

### Books DataFrame

```python
print('Books DataFrame - Head:')
display(books_df.head())
print('\nBooks DataFrame - Info:')
display(books_df.info())
print('\nBooks DataFrame - Describe:')
display(books_df.describe())
```

### Ratings DataFrame

```python
print('Ratings DataFrame - Head:')
display(ratings_df.head())
print('\nRatings DataFrame - Info:')
display(ratings_df.info())
print('\nRatings DataFrame - Describe:')
display(ratings_df.describe())
```

## 3. Exploratory Data Analysis (EDA)

### User Data Analysis

```python
# Distribution of Age
plt.figure(figsize=(10, 6))
sns.histplot(users_df['Age'], bins=30, kde=True)
plt.title('Distribution of User Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.75)
plt.show()

# Top 10 locations by user count
top_locations = users_df['Location'].value_counts().head(10)
plt.figure(figsize=(12, 6))
sns.barplot(x=top_locations.index, y=top_locations.values)
plt.title('Top 10 User Locations')
plt.xlabel('Location')
plt.ylabel('Number of Users')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
